## 2.1 Dasar Teori

Python menyediakan berbagai operator yang berfungsi untuk mengolah data dan variabel dalam sebuah program. Operator merupakan simbol atau kata khusus yang digunakan untuk melakukan perhitungan, perbandingan, pemberian nilai, penggabungan kondisi, serta pemeriksaan data. Pemahaman mengenai operator sangat penting karena penggunaannya hampir selalu ditemukan dalam proses pembuatan program.

1. Operator Aritmatika, digunakan untuk melakukan perhitungan matematika. Jenis operator ini meliputi penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus (%), dan perpangkatan (**). Operator tersebut dapat diterapkan pada data numerik seperti integer dan float, misalnya dalam menghitung nilai, luas bangun, jumlah transaksi, dan kebutuhan perhitungan lainnya.
2. Operator Perbandingan, berfungsi untuk membandingkan dua nilai. Hasil dari perbandingan berupa nilai Boolean, yaitu True atau False. Operator yang digunakan antara lain sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), dan lebih kecil atau sama dengan (<=). Operator ini sering digunakan untuk menentukan suatu kondisi dalam percabangan program.
3. Operator Logika, digunakan untuk menggabungkan atau memproses beberapa kondisi. Python memiliki tiga operator logika utama, yaitu and, or, dan not. Operator and menghasilkan nilai True apabila seluruh kondisi bernilai benar. Operator or menghasilkan nilai True apabila salah satu kondisi bernilai benar, sedangkan not berfungsi membalik nilai Boolean. Operator logika banyak digunakan dalam proses pengambilan keputusan.
4. Operator Penugasan, digunakan untuk memberikan nilai kepada variabel atau memperbarui nilai yang telah tersimpan. Operator dasar yang digunakan adalah =, sedangkan operator penugasan gabungan meliputi +=, -=, *=, /=, //=, %=, dan **=. Penggunaan operator ini dapat menyederhanakan penulisan program, terutama ketika nilai variabel perlu diubah berdasarkan nilai sebelumnya.
5. Operator Bitwise, digunakan untuk melakukan operasi pada tingkat bit dalam bilangan integer. Beberapa operator bitwise yang tersedia dalam Python adalah AND (&), OR (|), XOR (^), NOT (~), left shift (<<), dan right shift (>>). Operator ini bekerja dengan memproses bentuk biner suatu bilangan dan biasanya digunakan dalam pengolahan data tingkat rendah, sistem komputer, serta kebutuhan pemrograman tertentu.
6. Operator Keanggotaan, digunakan untuk memeriksa keberadaan suatu nilai di dalam objek data, seperti string, list, tuple, atau set. Python menyediakan operator in dan not in. Operator in menghasilkan True apabila nilai ditemukan dalam suatu objek, sedangkan not in menghasilkan True apabila nilai tidak ditemukan. Operator ini memudahkan proses pemeriksaan data dalam program.
7. Operator Identitas, digunakan untuk memeriksa apakah dua variabel mengacu pada objek yang sama di dalam memori. Operator yang digunakan adalah is dan is not. Operator is menghasilkan True apabila kedua variabel menunjuk pada objek yang sama, sedangkan is not menghasilkan True apabila keduanya menunjuk pada objek yang berbeda. Operator identitas berbeda dengan ==, karena == digunakan untuk membandingkan kesamaan nilai, bukan kesamaan objek.

### 2.2.1 Program aritmatika_2611532004

```python
# Buat file dengan nama aritmatika_2611532004.py
# Buat program untuk operator aritmatika dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_2004
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer 

angka1_2004 = int(input("Input angka-1: "))
angka2_2004 = int(input("Input angka-2: "))

# Penjumlahan
hasil = angka1_2004 + angka2_2004
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# Pengurangan
hasil = angka1_2004 - angka2_2004
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian
hasil = angka1_2004 * angka2_2004
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2004 != 0:
    hasil = angka1_2004 / angka2_2004
    print("\nOperator Pembagian")
    print ("Hasil =", hasil)

    hasil = angka1_2004 // angka2_2004
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_2004 % angka2_2004
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_2004 ** angka2_2004
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

### 2.2.2 Program assignment_2611532004

```python
# Buat file dengan assignment_2611532004.py
# Nama varibel ditambah 4 digit nim terakhir contoh: angka1_2004
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2004 = int(input("Input angka-1: "))
angka2_2004 = int(input("Input angka-2: "))

print("\nNilai awal angka1_2004 =", angka1_2004)
print("Nilai angka2_2004 =", angka2_2004)

# Assignment biasa
hasil = angka1_2004
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_2004
hasil += angka2_2004
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_2004
hasil -= angka2_2004
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_2004
hasil *= angka2_2004
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2004 != 0:
    hasil = angka1_2004
    hasil /= angka2_2004
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_2004
    hasil //= angka2_2004
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_2004
    hasil %= angka2_2004
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assingment perpangkatan
hasil = angka1_2004
hasil **= angka2_2004
print("\nAssignmnet perpangkatan (**=)")
print("Hasil =", hasil)
```

### 2.2.3 Program bitwise_2611532004
</div>

```python
# Buat file dengan nama bitwise_2611532004.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_2004
# Program ini menggunakan fungsi input()

print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================")

angka1_2004 = int(input("Masukkan angka bitwise-1: "))
angka2_2004 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_2004 =", "| biner =", bin(angka1_2004))
print("angka2_2004 =", "| biner =", bin(angka2_2004))

# Bitwise AND
hasil = angka1_2004 & angka2_2004
print("\nBitwise AND (&)")
print(angka1_2004, "&", angka2_2004, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise OR
hasil = angka1_2004 | angka2_2004
print("\nBitwise OR (|)")
print(angka1_2004, "|", angka2_2004, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise XOR
hasil = angka1_2004 ^ angka2_2004
print("\nBitwise XOR (^)")
print(angka1_2004, "^", angka2_2004, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise NOT
hasil = ~angka1_2004
print("\nBitwise NOT (~)")
print("~", angka1_2004, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kiri
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil = angka1_2004 << jumlah_geser
print("\nBitwise geser kiri (<<)")
print(angka1_2004, "<<", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kanan
hasil = angka1_2004 >> jumlah_geser
print("\nBitwise geser kanan (>>)")
print(angka1_2004, ">>", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))
```

```python
===================================
3. OPERATOR BITWISE
===================================

Angka dalam bentuk desimal dan biner
angka1_2004 = 10 | biner = 0b1010
angka2_2004 = 20 | biner = 0b10100

Bitwise AND (&)
10 & 20 = 0
Biner hasil = 0b0
Biner hasil (8 bit) = 00000000

Bitwise OR (|)
10 | 20 = 30
Biner hasil = 0b11110
Biner hasil (8 bit) = 00011110

Bitwise XOR (^)
10 ^ 20 = 30
Biner hasil = 0b11110
Biner hasil (8 bit) = 00011110

Bitwise NOT (~)
~ 10 = -11
Biner hasil = -0b1011
Biner hasil (8 bit) = -0001011

Bitwise geser kiri (<<)
10 << 1 = 20
Biner hasil = 0b10100
Biner hasil (8 bit) = 00010100

Bitwise geser kanan (>>)
10 >> 1 = 5
Biner hasil = 0b101
Biner hasil (8 bit) = 00000101
```

```python
# Buat file dengan nama lainnya_2611532004.py
# Nama varibel ditambah 4 digit nim terakhir contoh: angka1_2004
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data = [int(angka.strip()) for angka in input_data.split(",")]

nilai_dicari = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil = nilai_dicari in data
print("\nOperatorkeanggotaan IN")
print(nilai_dicari, data, "=", hasil)

# Operator not in
hasil = nilai_dicari not in data
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data, "=", hasil)


print("\n===================================")
print("2. OPERATOR IDENTITAS")
print("===================================")

# objek1 menggunakan list dari input pengguna
objek1 = data

# objek2 merujuk pada objek yang sama dengan objek1
objek2 = objek1

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3 = data.copy()

print("objek1 =", objek1)
print("objek2 =", objek2)
print("objek3 =", objek3)

# Operator is
hasil = objek1 is objek2
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil)

# Operator is not
hasil = objek1 is not objek3
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas nilai")
print("objek1 is objek3 =", objek1 is objek3)
print("objek1 == objek3 =", objek1 == objek3)
```

### 2.2.5 Program logika_2611532004
</div>

```python
# Buat file dengan nama logika_2611532004.py
# Nama varibel ditambah 4 digit nim terakhir contoh: a1_2004
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2004 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2004 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_2004)
print("A2 =", a2_2004)

# Konjungsi: bernilai True jika keduanya True
hasil = a1_2004 and a2_2004 
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_2004 or a2_2004
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_2004
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_2004
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_2004 != a2_2004
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2 =", hasil)
```

```python
A1 = True
A2 = False

Konjungsi (AND)
A1 and A2 = False

Disjungsi (OR)
A1 or A2 = True

Negasi A1 (NOT)
not A1 = False

Negasi A2 (NOT)
not A2 = True

Disjungsi Ekslusif (XOR)
A1 XOR A2 = True
```

### 2.2.6 Program perbandingan_2611532004

```python
# Buat file dengan nama perbandingan_2611532004.py
# Nama varibel ditambah 4 digit nim terakhir contoh: angka1_2004
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2004 = int(input("Input angka-1: "))
angka2_2004 = int(input("Input angka-2: "))

# Lebih besar dari
hasil = angka1_2004 > angka2_2004
print("\nOperator lebih besar dari")
print("angka1_2004 > angka2_2004 =", hasil)

# Lebih kecil dari
hasil = angka1_2004 < angka2_2004
print("\nOperator lebih kecil dari")
print("angka1_2004 < angka2_2004 =", hasil)

# Lebih besar dari atau sama dengan
hasil = angka1_2004 >= angka2_2004
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_2004 >= angka2_2004 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_2004 <= angka2_2004
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_2004 <= angka2_2004 =", hasil)

# Sama dengan
hasil = angka1_2004 == angka2_2004
print("\nOperator sama dengan")
print("angka1_2004 == angka2_2004 =", hasil)

# Tidak sama dengan
hasil = angka1_2004 != angka2_2004
print("\nOperator tidak sama dengan")
print("angka1_2004 != angka2_2004 =", hasil)

# Tambahan: perbandingan berantai dalam Python
hasil = 0 < angka1_2004 < 100
print("\nPerbandingan berantai")
print("0 , angka1_2004 < 100 =", hasil)

hasil = 0 < angka2_2004 < 100
print("0 < angka2_2004 < 100 =", hasil)
```

```python
Operator lebih besar dari
angka1_2004 > angka2_2004 = False

Operator lebih kecil dari
angka1_2004 < angka2_2004 = True

Operator lebih besar dari atau sama dengan
angka1_2004 >= angka2_2004 = False

Operator lebih kecil dari atau sama dengan
angka1_2004 <= angka2_2004 = True

Operator sama dengan
angka1_2004 == angka2_2004 = False

Operator tidak sama dengan
angka1_2004 != angka2_2004 = True

Perbandingan berantai
0 , angka1_2004 < 100 = True
0 < angka2_2004 < 100 = True
```
