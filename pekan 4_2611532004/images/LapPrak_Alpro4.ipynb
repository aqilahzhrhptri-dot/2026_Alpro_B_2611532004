## 2.1 Dasar Teori

Conditional merupakan struktur dalam Python yang digunakan untuk menjalankan perintah berdasarkan kondisi tertentu. Dengan conditional, program dapat menentukan apakah suatu perintah akan dijalankan atau tidak berdasarkan hasil evaluasi kondisi.

1. Pernyataan if
Pernyataan if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). umur = 18.
2. Pernyataan if-else
Pernyataan if-else digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada if bernilai benar, maka perintah pada if dijalankan. Jika salah, program menjalankan perintah pada else.
3. Pernyataan if-elif-else
Pernyataan if-elif-else digunakan ketika terdapat lebih dari dua kondisi yang perlu diperiksa. Python akan memeriksa kondisi secara berurutan hingga menemukan kondisi yang bernilai benar.
4. Pernyataan match-case
Pernyataan match-case digunakan untuk memeriksa suatu nilai terhadap beberapa kemungkinan pola (case) sekaligus. Python akan mencocokkan nilai dengan case yang sesuai, dan case _ digunakan untuk menangani nilai yang tidak cocok dengan case manapun.

### 2.2.4 Program multi_if2_2611532004
</div>

```python
# Buat file dengan nama multi_if2_26115320004.py
# Buat program untuk kondisional if
# Nama variabel ditambaha 4 digit nim terakihir contoh: total belanja_2004
# Program ini manggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2004 = float(input("Masukkan total belanja (Rp): "))

# Input status member(mengecek apakah user mengecek 'y' atau 't')
input_member_2004 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_2004 = input_member_2004 in ["y", "t"]

# Input status kode promo (mengecek apakah user mengetik 'y atau t')
input_promo_2004 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2004 = input_promo_2004 in ["y", "t"]

total_diskon_persen_2004 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen
# Disko bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2004 > 1000000: 
    total_diskon_persen_2004 += 10 # Diskon belanja besar

if is_member_2004:
    total_diskon_persen_2004 += 5 # Diskon member

if kode_promo_valid_2004:
    total_diskon_persen_2004 += 15 # Diskon voucher 

# Menghitung nominal diskon dan total bayar
nominal_diskon_2004 = total_belanja_2004 * (total_diskon_persen_2004 / 100)
total_bayar_2004 = total_belanja_2004 - nominal_diskon_2004

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Totsl Diskon : {total_diskon_persen_2004}% (Rp {nominal_diskon_2004:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2004:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2004}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Program meminta pengguna memasukkan total belanja menggunakan input(), kemudian diubah ke float dan disimpan pada variabel total_belanja_2004.
2. Program meminta status keanggotaan dengan mengetik 'y' atau 't'. Input tersebut dirapikan dengan .strip().lower(), lalu dicek apakah hasilnya termasuk dalam ["y", "t"] dan disimpan sebagai nilai boolean pada is_member_2004.
3. Program memeriksa validitas kode promo dan menyimpannya sebagai nilai boolean pada kode_promo_valid_2004.
4. Program menyiapkan variabel total_diskon_persen_2004 dengan nilai awal 0 untuk menampung akumulasi diskon.
5. Program menggunakan multi-if terpisah, yaitu tiga pernyataan if yang diperiksa secara independen, bukan elif, sehingga diskon dapat ditumpuk jika beberapa syarat terpenuhi sekaligus.
6. if pertama memeriksa apakah total belanja lebih dari Rp1.000.000, jika benar maka total_diskon_persen_2004 ditambah 10.
7. if kedua memeriksa status member, jika benar maka total_diskon_persen_2004 ditambah 5.
8. if ketiga memeriksa validitas kode promo, jika benar maka total_diskon_persen_2004 ditambah 15.
9. Program menghitung nominal_diskon_2004 (nilai potongan dalam rupiah) dan total_bayar_2004 (total belanja dikurangi nominal diskon).

### 2.2.7 Program match_case1_2611532004
</div>

```python
# BUat file dengan nama match_case1_2611532004.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2004
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan_2004 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2004:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Program meminta pengguna memasukkan angka bulan (1–12) menggunakan input(), kemudian diubah menjadi int dan disimpan pada variabel bulan_2004.
2. Program menggunakan struktur match-case untuk mengonversi angka bulan menjadi nama bulan.
3. Nilai bulan_2004 dicocokkan dengan setiap case yang tersedia, mulai dari case 1 hingga case 12, di mana masing-masing mewakili satu nama bulan (Januari sampai Desember).
4. Jika bulan_2004 cocok dengan salah satu case tersebut, program mencetak nama bulan yang sesuai.
5. Jika bulan_2004 tidak cocok dengan angka 1–12 (misalnya 0 atau 13), program menjalankan case _ sebagai penampung kondisi default, yaitu mencetak "Angka tidak valid".
6. Karena match-case hanya mencocokkan satu case yang sesuai, hanya satu output saja yang tercetak sesuai angka bulan yang diinput.

```python
import sys

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_pengunjung_2004 = input("Masukkan Nama Pengunjung :")
umur_2004 = int(input("Input Umur Anda :"))
sim_2004 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba          (Rp 50,000)")
print(" 2. Arung Jeram           (Rp 75,000)")
print(" 3. Motor ATV Ekstrim     (Rp 120,000)")
print(" 4. Roller Coaster Kilat  (Rp 100,000)")
print(" 5. All-Access VIP        (RP 220,000)")
paket_2004 = int(input("Masukkan nomor paket (1-5) :"))

match paket_2004:
    case 1:
        nama_wahana_2004 = "Wahana Safari Rimba"
        harga_satuan_2004 = 50000
    case 2:
        nama_wahana_2004 = "Wahana Arung Jeram"
        harga_satuan_2004 = 75000
    case 3:
        nama_wahana_2004 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2004 = 120000
    case 4:
        nama_wahana_2004 = "Wahana Roller Coaster Kilat"
        harga_satuan_2004 = 100000
    case 5:
        nama_wahana_2004 = "Wahana All-Access VIP"
        harga_satuan_2004 = 220000
    case _:
        print("Paket wahana tidak valid!")
        sys.exit()

jumlah_tiket_2004 = int(input("Masukkan jumlah tiket :"))
if jumlah_tiket_2004 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")
is_member_2004 = input("Apakah Anda member? (y/t) : ").strip().lower()
kode_promo_valid_2004 = input("Apakah kode promo valid? (y/t) : ").strip().lower()

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if paket_2004 == 3 and umur_2004 >= 17 and sim_2004 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_2004 == 3 and umur_2004 >= 17 and sim_2004 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_2004 == 3 and umur_2004 < 17 and sim_2004 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_2004 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif umur_2004 >= 10:
    print("Status Akses: Anda memenuhi syarat umur untuk memiliki wahana ini.")
else:
    print("Status Akses: Anda belum memenuhi syarat umur untuk menaiki wahana ini.")

subtotal_2004 = harga_satuan_2004 * jumlah_tiket_2004
total_diskon_persen_2004 = 0
if subtotal_2004 >= 200000:
    total_diskon_persen_2004 += 10 # Diskon Belanja Besar
if is_member_2004 in ['y', 'ya']:
    total_diskon_persen_2004 += 5 # Diskon member
if kode_promo_valid_2004 in ['y', 'ya']:
    total_diskon_persen_2004 += 15 # Diskon Voucher Promo
if jumlah_tiket_2004 >= 5:
    total_diskon_persen_2004 += 5 # Diskon Tambahan Rombongan
nominal_diskon_2004 = subtotal_2004 * (total_diskon_persen_2004 / 100)
total_bayar_2004 = subtotal_2004 - nominal_diskon_2004
if total_bayar_2004 > 300000:
    catatan_layanan_2004 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_2004 = "Terimakasih telah berkunjung"

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_2004:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_2004}% (RP {nominal_diskon_2004:,.0f})")
print(f"Total Bayar      :Rp {total_bayar_2004:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_2004}")
print("Program Selesai")
```

Jika angka yang diinput tidak ada dalam pilihan 1–5, case _ akan mencetak "Paket wahana tidak valid!" dan menghentikan program dengan sys.exit().

Program menghitung subtotal_2004 (harga satuan dikali jumlah tiket), lalu menggunakan multi-if terpisah — empat pernyataan if independen — untuk mengakumulasikan total_diskon_persen_2004 berdasarkan subtotal besar, status member, kode promo, dan jumlah tiket rombongan, sehingga diskon bisa ditumpuk jika beberapa syarat terpenuhi sekaligus.
