Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub ini memuat pembahasan mendalam mengenai fondasi lingkungan kerja pemrogram modern yang mencakup aspek bahasa pemrograman, editor, serta sistem kontrol versi.

### 2.2.2 Membuat Repository GitHub
Setelah aplikasi selesai dibuat dan diuji menggunakan VS Code, langkah selanjutnya adalah mengintegrasikan proyek tersebut dengan GitHub. Repository berfungsi sebagai tempat penyimpanan kode program secara online.

Langkah-langkah yang perlu dilakukan adalah sebagai berikut:

1. Membuat repository baru di GitHub.
2. Membuka folder proyek melalui VS Code.
3. Mengaktifkan terminal di dalam VS Code.
4. Melakukan inisialisasi Git pada folder proyek tersebut.
5. Memasukkan file-file proyek ke tahap staging.
6. Membuat commit sebagai catatan perubahan.
7. Menautkan repository lokal dengan repository yang ada di GitHub.
8. Melakukan push untuk mengunggah kode program ke GitHub.

Contoh perintah yang digunakan:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:
1. git init berfungsi untuk menginisialisasi atau membuat repository Git baru di dalam folder proyek.
2. git add . berfungsi untuk memasukkan seluruh file yang telah diubah atau ditambahkan ke dalam staging area.
3. git commit berfungsi untuk merekam atau menyimpan perubahan yang telah di-staging, disertai dengan pesan sebagai keterangan.
4. git branch -M main berfungsi untuk mengganti atau menetapkan nama branch utama menjadi "main".
5. git remote add origin berfungsi untuk menautkan repository lokal dengan repository yang berada di GitHub.
6. git push berfungsi untuk mengirim atau mengunggah kode dari komputer lokal ke repository GitHub.
