## 2.1 Dasar Teori
1. Variabel dan Input Pengguna. Variabel adalah tempat untuk menyimpan data di dalam memori selama program berjalan. Di Python, variabel tidak perlu dideklarasikan tipenya terlebih dahulu. Tipe data ditentukan otomatis dari nilai yang diberikan, misalnya bilangan bulat (int), bilangan desimal (float), dan teks (str).
Data dari pengguna dapat diterima melalui fungsi input(). Fungsi ini selalu menghasilkan data bertipe teks, sehingga jika dibutuhkan dalam perhitungan, nilainya harus diubah dulu, misalnya dengan int().

2. Operator aritmatika. Digunakan untuk melakukan perhitungan matematika dasar. Pada praktikum ini, operator tersebut dipakai untuk menghitung jumlah spasi dan karakter pada setiap baris pola, menaikkan nilai penghitung, serta menentukan bilangan genap atau ganjil dengan operator %.
- (+) untuk penjumlahan
- (-) untuk pengurangan
- (*) untuk perkalian
- (%) (modulus) untuk menghasilkan sisa bagi, misalnya 7 % 2 menghasilkan 1. Operator ini sering dipakai untuk menentukan bilangan genap atau ganjil.

3. Operator perbandingan. Digunakan untuk membandingkan dua nilai dan menghasilkan True atau False. Hasil tersebut dipakai sebagai syarat pada percabangan if, elif, dan else, misalnya untuk memeriksa apakah suatu angka genap (angka % 2 == 0) atau untuk menentukan batas pada pola.
Hasil dari operator perbandingan berupa nilai True atau False.
- (<) lebih kecil dari
- (>) lebih besar dari
- (<=) lebih kecil dari atau sama dengan
- (==) sama dengan
- (!=) tidak sama dengan

4. Percabangan if, elif, dan else. Percabangan digunakan agar program dapat memilih tindakan berdasarkan suatu kondisi. Pernyataan if dijalankan apabila kondisinya bernilai benar. Pernyataan elif digunakan untuk memeriksa kondisi lain apabila kondisi sebelumnya tidak terpenuhi. Pernyataan else dijalankan apabila semua kondisi sebelumnya tidak terpenuhi.

5. Perulangan for. Perulangan atau looping adalah struktur yang menjalankan sekumpulan instruksi secara berulang. Perulangan for di Python bekerja dengan mengambil nilai satu per satu dari sebuah urutan, lalu menjalankan blok kode untuk setiap nilai tersebut.

6. Fungsi range(). Fungsi range() berfungsi membuat deretan bilangan yang biasanya dipakai sebagai acuan dalam perulangan for. Melalui fungsi ini, programmer dapat mengatur angka pertama yang dijadikan titik mulai, angka yang menjadi batas berhentinya perulangan, dan besar kenaikan atau penurunan nilai di setiap putarannya.
Penulisan range() dapat dilakukan dalam tiga bentuk berikut.
- range(batas)
- range(awal, batas)
- range(awal, batas, langkah)
Pada praktikum ini, nilai awal perulangan disesuaikan dengan kebutuhan program, ada yang dimulai dari 0 dan ada yang dari 1.

7. Nested for. Nested for adalah bentuk perulangan for yang diletakkan di dalam perulangan for yang lain. Dengan susunan ini, program dapat menjalankan proses berulang secara berlapis, sehingga cocok dipakai ketika hasil yang diinginkan perlu ditata dalam bentuk baris dan kolom.
Dalam praktikum ini, nested for dipakai untuk membuat beragam keluaran, di antaranya pola bintang, pola titik, deret angka, segitiga, dan jam pasir. Perulangan luar umumnya bertugas menentukan banyaknya baris atau posisi baris yang sedang diproses, sementara perulangan dalam bertugas mengatur karakter atau angka yang dicetak pada tiap barisnya.

8. Parameter end pada print(). Secara bawaan, print() di Python akan berpindah ke baris baru setelah menampilkan sebuah nilai. Agar tampilan tidak langsung turun ke baris berikutnya, parameter end dapat diatur sesuai kebutuhan.
Penulisan end="" membuat cetakan selanjutnya tetap berada di baris yang sama dan tidak ada karakter tambahan di antaranya. Adapun end=" " juga mempertahankan keluaran dalam satu baris, tetapi menambahkan satu spasi setelah nilai yang dicetak.
Dalam praktikum ini, pengaturan tersebut dipakai ketika angka, karakter, dan spasi perlu disusun sedikit demi sedikit dalam satu baris, sehingga pola yang dihasilkan sesuai dengan bentuk yang diharapkan.

9. Pencetakan Pola. Pola pada Python dapat dibuat dengan nested for yang mengatur susunan karakter atau angka. Bentuk akhirnya bergantung pada beberapa hal, yaitu berapa banyak karakter yang dicetak, berapa jumlah spasi di depannya, di mana angka diletakkan, dan kapan program harus pindah ke baris berikutnya.
Pada praktikum ini, nested for dipakai untuk membuat beberapa jenis pola, di antaranya susunan titik dan angka, persegi dari karakter *, pola yang dibentuk dari hasil perhitungan i + j, pola jam pasir, dan pola segitiga. Dari latihan tersebut dapat dilihat bahwa setiap bagian pola ditentukan oleh perulangan yang mengatur baris dan kolomnya.

10. Pola jam pasir kristal palindromik. Merupakan gabungan dari beberapa konsep yang sudah dipelajari, yaitu perulangan, percabangan, perhitungan, dan pencetakan karakter. Pola ini disusun dalam tiga bagian: bagian atas, bagian tengah yang menjadi poros, dan bagian bawah.
Pada tiap baris, angka dicetak dari nilai terbesar turun sampai 1, lalu naik lagi dari 1 sampai nilai terbesar. Susunan inilah yang membuatnya disebut palindromik, karena terbaca sama dari kiri maupun kanan. Di bagian tengah ditempatkan simbol <*> sebagai pusat pola, sedangkan karakter #, =, dan | dipakai untuk membuat bingkai di sekelilingnya.
Jumlah spasi di sisi kiri dan kanan berubah mengikuti nomor barisnya. Jika perhitungannya tepat, setiap baris akan berada di posisi yang sesuai dan bentuk jam pasir yang dihasilkan terlihat simetris.

### 2.2.1 Program jumlah_genap_2611532004
</div>

```python
# Buat file dengan nama jumlah_genap_2004.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_2004
# Program ini menggunakan fungsi input()

ulang_2004 = int(input("Masukkan nilai batas: "))

jumlah_2004 = 0
for i_2004 in range(1, ulang_2004 + 1):
    if i_2004 % 2 == 0:
        print(i_2004, end=" ")
        jumlah_2004 = jumlah_2004 + i_2004

        if i_2004 < ulang_2004:
            print(" + ", end="")
        else:
            print(" = ", jumlah_2004, end="")
print()
print("Jumlah =", jumlah_2004)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. ulang_2004 = int(input("Masukkan nilai batas: ")) meminta pengguna memasukkan angka batas. Nilainya diubah ke bilangan bulat dengan int() agar bisa dipakai dalam perhitungan.
2. jumlah_2004 = 0 membuat variabel penampung jumlah dengan nilai awal 0.
3. for i_2004 in range(1, ulang_2004 + 1) mengulang angka dari 1 sampai batas. Ditambah 1 supaya angka batas ikut diperiksa.
4. if i_2004 % 2 == 0 memeriksa apakah angka genap, yaitu angka yang sisa baginya dengan 2 adalah 0.
5. print(i_2004, end=" ") mencetak angka genap tanpa pindah baris.
6. jumlah_2004 = jumlah_2004 + i_2004 menambahkan angka genap tersebut ke jumlah.
7. if i_2004 < ulang_2004 mencetak tanda + jika angka belum mencapai batas, sedangkan else mencetak tanda = beserta hasil jumlahnya.
8. print() pindah ke baris baru, lalu print("Jumlah =", jumlah_2004) menampilkan total jumlah bilangan genap.


### 2.2.2 Program nested_for1_2611532004

```python
# Buat file dengan nama nested_for1_2611532004.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_2004
# Program ini menggunakan fungsi input()

batas_2004 = int(input("Masukkan nilai batas: "))
for line_2004 in range(1, batas_2004 + 1):
    for j_2004 in range(1, (-1 * line_2004 + batas_2004) + 1):
        print(" . ", end="")
    print(line_2004)
```

### 2.2.5 Program nested_for4_2611532004

```python
# Buat file dengan nama nested_for4_2611532004.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_2004
# Program ini menggunakan fungsi input()

tinggi_2004 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2004 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2004 = tinggi_2004
    c_2004 = a_2004
    lebar_2004 = (2 * tinggi_2004 + 1)

    for i_2004 in range(1, tinggi_2004 + 1):
        b_2004 = c_2004 + 1

        for j_2004 in range(1, lebar_2004 + 1):


            

            # Baris atas dan bawah
            if i_2004 == 1 or i_2004 == tinggi_2004:
                if j_2004 == 1 or j_2004 == lebar_2004:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2004 == 1 or j_2004 == lebar_2004:
                    print("|", end="")
                else:
                    if j_2004 == c_2004:
                        print("<", end="")
                    elif j_2004 == b_2004:
                        print(">", end="")
                    elif j_2004 == (lebar_2004 - c_2004):
                        print("<", end="")
                    elif j_2004 == (lebar_2004  - c_2004 + 1):
                        print(">", end="")
                    elif j_2004 > b_2004 and j_2004 < (lebar_2004 - c_2004):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_2004 -= 2

        if a_2004 <= 0:
            c_2004 = (-a_2004) + 2
        else:
            c_2004 = a_2004
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. tinggi_2004 = int(input(...)) meminta pengguna memasukkan tinggi pola, lalu nilainya diubah menjadi bilangan bulat dengan int().

3. a_2004, c_2004, dan lebar_2004 disiapkan sebagai variabel awal. a_2004 dan c_2004 bernilai sama dengan tinggi, dengan c_2004 menjadi posisi kolom simbol < di sisi kiri. Lebar pola dihitung dengan rumus 2 * tinggi_2004 + 1, sehingga bernilai 21 jika tingginya 10.
4. for i_2004 in range(1, tinggi_2004 + 1) adalah perulangan luar yang mengatur baris. Di awal setiap baris, b_2004 = c_2004 + 1 dihitung sebagai posisi simbol > yang berada tepat di sebelah <.
5. for j_2004 in range(1, lebar_2004 + 1) adalah perulangan dalam yang mengatur kolom. Setiap karakter dicetak satu per satu dengan end="" agar tetap berada dalam satu baris.
6. Baris pertama dan terakhir (i_2004 == 1 or i_2004 == tinggi_2004) dicetak sebagai bingkai atas dan bawah. Kolom paling kiri dan paling kanan diisi #, sedangkan kolom di antaranya diisi =.
7. Pada baris isi, kolom paling kiri dan paling kanan diisi |. Kolom di antaranya ditentukan oleh if, elif, dan else: kolom c_2004 dan lebar_2004 - c_2004 dicetak <, kolom b_2004 dan lebar_2004 - c_2004 + 1 dicetak >, kolom di antara > kiri dan < kanan dicetak ., dan kolom sisanya dicetak spasi.
8. print() setelah perulangan dalam berfungsi pindah ke baris berikutnya.
9. Bagian akhir memperbarui posisi <> untuk baris selanjutnya. a_2004 -= 2 mengurangi nilai a_2004 sebanyak 2 di setiap baris. Selama a_2004 masih lebih dari 0, c_2004 mengikuti nilai a_2004 sehingga <> bergeser ke kiri. Setelah a_2004 bernilai 0 atau kurang, c_2004 dihitung dengan (-a_2004) + 2 sehingga <> bergeser kembali ke tengah. Inilah yang membuat pola simetris dari atas ke bawah.

```python
# Buat file dengan nama perulangan_for3_2611532004.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_2004
# Program ini menggunakan fungsi input()

ulang_2004 = int(input("Masukkan jumlah perulangan: "))

jumlah_2004 = 0
for i_2004 in range(1, ulang_2004 + 1):
    print(i_2004, end=" ")
    jumlah_2004 = jumlah_2004 + i_2004

    if i_2004 < ulang_2004:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2004, end="")
print()
print("Jumlah =", jumlah_2004)
```

2. jumlah_2004 = 0 membuat variabel penampung jumlah dengan nilai awal 0.
3. for i_2004 in range(1, ulang_2004 + 1) mengulang angka dari 1 sampai batas. Ditambah 1 supaya angka batas ikut diproses.
4. print(i_2004, end=" ") mencetak angka pada putaran tersebut tanpa pindah baris.
5. jumlah_2004 = jumlah_2004 + i_2004 menambahkan angka tersebut ke jumlah.
6. if i_2004 < ulang_2004 mencetak tanda + jika angka belum mencapai batas, sedangkan else mencetak tanda = beserta hasil jumlahnya pada putaran terakhir.
7. print() pindah ke baris baru, lalu print("Jumlah =", jumlah_2004) menampilkan total penjumlahan.

```python
# Program perulangan for untuk membuat segitiga

tinggi_2004 = int(input("Masukkan tinggi segitiga: "))
for i_2004 in range(1, tinggi_2004 + 1):
    for j_2004 in range(1, tinggi_2004 - i_2004 + 1):
        print(" ", end=" ")
    for k_2004 in range(1, 2 * i_2004):
        print("*", end=" ")
    print()
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

4. print(" ", end=" ") mencetak satu spasi ditambah satu spasi dari end, sehingga setiap putaran menghasilkan dua spasi. Lebar ini sama dengan lebar "* " pada bintang, supaya bentuknya tetap lurus.
5. for k_2004 in range(1, 2 * i_2004) adalah perulangan dalam kedua yang mencetak bintang. Jumlah putarannya 2 * i_2004 - 1, yaitu 1, 3, 5, 7, dan seterusnya, sehingga setiap baris bertambah dua bintang.
6. print("*", end=" ") mencetak satu bintang yang diikuti spasi. Parameter end=" " membuat bintang berikutnya tetap berada di baris yang sama.
7. print() dijalankan setelah kedua perulangan dalam selesai. Perintah ini pindah ke baris baru, sehingga baris berikutnya tercetak di bawahnya.


### Tugas Praktikum

Studi kasus ini meminta mahasiswa membuat program Python interaktif yang menerima bilangan bulat positif N, lalu mencetak pola jam pasir kristal palindromik di dalam bingkai berkarakter #, =, dan |, yang terdiri dari tiga fase: bagian atas dengan baris N turun sampai 1, satu baris poros pusat <*>, dan bagian bawah dengan baris 1 naik sampai N, di mana setiap barisnya memuat deret angka mundur, poros <*>, dan deret angka maju yang diseimbangkan dengan spasi pada sisi kiri dan kanan, dengan ketentuan semua variabel berakhiran empat digit terakhir NIM, seluruh bingkai, spasi, dan angka dicetak menggunakan nested for dengan end="" tanpa perkalian string, serta setiap akhir baris ditutup dengan print() tanpa argumen.


#### Kode Jawaban 
<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_2004 = int(input("Masukkan ukuran skala jam pasir (N): "))

print("#", end="")
for garis_2004 in range(4 * n_2004 + 5):
    print("=", end="")
print("#", end="")
print()

for baris_2004 in range(n_2004, 0, -1):
    print("| ", end="")
    for spasi_kiri_2004 in range(2 * (n_2004 - baris_2004)):
        print(" ", end="")
    for angka_2004 in range(baris_2004, 0, -1):
        print(angka_2004, end=" ")
    print("<*>", end="")
    for angka_2004 in range(1, baris_2004 + 1):
        print(" ", end="")
        print(angka_2004, end="")
    for spasi_kanan_2004 in range(2 * (n_2004 - baris_2004)):
        print(" ", end="")
    print(" |", end="")
    print()

print("|", end="")
for spasi_kiri_2004 in range(2 * n_2004 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanan_2004 in range(2 * n_2004 + 1):
    print(" ", end="")
print("|", end="")
print()

for baris_2004 in range(1, n_2004 + 1):
    print("| ", end="")
    for spasi_kiri_2004 in range(2 * (n_2004 - baris_2004)):
        print(" ", end="")
    for angka_2004 in range(baris_2004, 0, -1):
        print(angka_2004, end=" ")
    print("<*>", end="")
    for angka_2004 in range(1, baris_2004 + 1):
        print(" ", end="")
        print(angka_2004, end="")
    for spasi_kanan_2004 in range(2 * (n_2004 - baris_2004)):
        print(" ", end="")
    print(" |", end="")
    print()

print("#", end="")
for garis_2004 in range(4 * n_2004 + 5):
    print("=", end="")
print("#", end="")
print()
```

Penjelasan:
1. print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===") menampilkan judul program.
2. n_2004 = int(input("Masukkan ukuran skala jam pasir (N): ")) meminta pengguna memasukkan ukuran N. Nilainya diubah ke bilangan bulat dengan int() agar bisa dipakai dalam range().
3. Bingkai atas dibuat dengan mencetak #, lalu perulangan for garis_2004 in range(4 * n_2004 + 5) mencetak = sebanyak 4N + 5, kemudian ditutup #. print() pindah ke baris baru. Bingkai bawah di akhir program memakai cara yang sama.
4. Fase 1 (jam pasir atas) memakai for baris_2004 in range(n_2004, 0, -1), sehingga nomor baris berjalan turun dari N sampai 1. Setiap baris diawali "| " yaitu garis tegak dan satu spasi padding.
5. Spasi penyeimbang kiri dicetak dengan for spasi_kiri_2004 in range(2 * (n_2004 - baris_2004)). Pada baris pertama jumlahnya 0, lalu bertambah 2 di setiap baris berikutnya sehingga bentuknya menyempit ke tengah.
6. Angka mundur dicetak dengan for angka_2004 in range(baris_2004, 0, -1) dan print(angka_2004, end=" "). Angka berjalan dari nomor baris turun ke 1, masing-masing diikuti satu spasi.
7. print("<*>", end="") mencetak poros kristal di tengah baris.
8. Angka maju dicetak dengan for angka_2004 in range(1, baris_2004 + 1). Di setiap putaran, program mencetak satu spasi lalu angkanya, sehingga deret naik dari 1 sampai nomor baris. Dengan begitu deret kiri dan kanan membentuk susunan palindromik.
9. Spasi penyeimbang kanan dicetak dengan rumus yang sama seperti sisi kiri, yaitu 2 * (n_2004 - baris_2004). Setelah itu print(" |", end="") mencetak padding dan garis tegak kanan, lalu print() pindah baris.
10. Fase 2 (poros pusat) hanya satu baris. Program mencetak |, spasi sebanyak 2 * n_2004 + 1, <*>, spasi dengan jumlah yang sama, lalu |.
